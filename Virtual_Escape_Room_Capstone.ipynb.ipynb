```python
# Capstone Project: Virtual Escape Room
# The Disappearance of Mr. White
# A story-driven Python game where the player solves randomized
# puzzles to uncover the mystery of Mr. White's disappearance
# before the 5-minute timer expires.

import random
import time

```

```python
time_limit = 300
start_time = time.time()
```

```python



print("🔎 THE DISAPPEARANCE OF MR. WHITE 🔎")
print()
print("You are a detective investigating the mysterious disappearance of Mr. White.")
print("The wealthy businessman vanished from his mansion without a trace.")
print("Hidden clues and puzzles throughout the mansion may reveal what happened.")
print()
print("You have only 5 minutes to solve the mystery. Good luck, Detective!")
```

```python
puzzles = [
    {
        "question": "I have keys but no locks. I have space but no room. You can enter, but you cannot go outside. What am I?",
        "answer": "keyboard",
        "story": "The desk opens! Inside, you discover evidence that Mr. White was planning a secret meeting."
        },
    {
        "question": "The secret meeting was scheduled for 20:00. What hour is that on a 12-hour clock?",
        "answer": "8",
        "story": "You discover evidence showing that Mr. White planned to meet someone at 8 PM on the night he disappeared."
    },
    {
        "question": "In the library, four books show the pattern: 2, 4, 8, 16, ?. What number comes next?",
        "answer": "32",
        "story": "A hidden compartment opens! Inside is a photograph of Mr. White with an unknown person."
    },
    {
        "question": "The basement keypad asks: What hour was Mr. White's secret meeting?",
        "answer": "8",
        "story": "The basement door opens! Inside, you discover evidence suggesting that Mr. White's disappearance may have been planned."
    }
]
```

```python
random.shuffle(puzzles)
```

```python
def check_answer(user_answer, correct_answer):
    if user_answer.lower().strip() == correct_answer.lower():
        return True
    else:
        return False
```

```python
completed_puzzles = 0

for puzzle in puzzles:
    elapsed_time = time.time() - start_time
    remaining_time = time_limit - elapsed_time

    if remaining_time <= 0:
        print("⏰ TIME'S UP!")
        print("Mr. White's disappearance remains a mystery...")
        break

    print()
    print("🔐 PUZZLE")
    print(puzzle["question"])

    solved = False

    while not solved:
        if remaining_time <= 0:
            break

        elapsed_time = time.time() - start_time
        remaining_time = time_limit - elapsed_time

        if remaining_time <= 0:
            print("⏰ TIME'S UP!")
            print("Mr. White's disappearance remains a mystery...")
            break

        user_answer = input("Your answer: ")

        if check_answer(user_answer, puzzle["answer"]):
            print("✅ Correct!")
            print(puzzle["story"])
            solved = True
            completed_puzzles += 1
        else:
            print("❌ Incorrect! 15 seconds have been deducted.")
            start_time -= 15

    if completed_puzzles == len(puzzles):
        print()
        print("🎉 CASE SOLVED! 🎉")
        print("You solved every puzzle before time ran out.")
        print("The evidence reveals that Mr. White planned his own disappearance.")
        print("He staged the mystery to escape his old life after discovering that someone close to him was plotting against him.")
        print("You found the final evidence in the basement and uncovered the truth!")
        print()
        print("🔎 Congratulations, Detective!")
```
